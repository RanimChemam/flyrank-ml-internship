## 1. My lane as an ML task (type)

I frame this lane as a ranking task.

The goal is to rank content pages by priority for human review, rather than making an automatic yes/no decision. The output would be an ordered list of pages so that an SEO or content specialist can review the highest-priority pages first.

Ranking fits the decision because the practical question is not only whether a page has an issue, but which pages should receive attention before others.

```python
print("ML task type: Ranking")
print("Unit of analysis: one content item/page")
print("Output: ranked priority list for human review")
```

## 2. Target or proxy

I will use the observed `trend_direction` as a proxy for content performance direction.

The dataset provides this field as an observed label derived from recent performance signals. I can use it to investigate whether the available content and search signals can help rank pages that deserve attention.

This is a proxy rather than a future outcome. It does not prove that a page will continue to decline or that updating it will improve performance.

```python
import pandas as pd

df = pd.read_csv(
    "/content/flyrank-ml-internship/data/raw/content_refresh_anonymized.csv"
)

print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
print("Proxy column: trend_direction")
print("Unique values:", df["trend_direction"].unique())
print("Missing values:", df["trend_direction"].isna().sum())
```

## 3. Success metric

I will use Precision@50 as the primary success metric.

Precision@50 measures how many of the 50 highest-ranked pages match the selected target or proxy. This metric fits the decision because the practical action is to give a human specialist a short list of pages to review first.

A higher Precision@50 means that the top of the ranking contains more relevant pages. I will compare this metric against a simple baseline rather than treating any single score as proof that the ranking is useful.

```python
print("Primary success metric: Precision@50")
print("Evaluation focus: the top 50 ranked pages")
print("Higher Precision@50 is better.")
```

## 4. The unit of analysis, as a real dataframe

The unit of analysis is one content item/page.

Each row represents one content item identified by `content_id`. The features describe its search demand, traffic, engagement, content age, freshness, and recent performance signals.

The decision is made at the page level, because the output will be used to prioritize individual pages for human review.

```python
page_sample = df[
    [
        "content_id",
        "client_id",
        "search_volume",
        "impressions_90d",
        "clicks_90d",
        "sessions_90d",
        "content_age_days",
        "days_since_last_update",
        "trend_direction"
    ]
].head(10)

page_sample
```

## 5. Why ML beats a fixed rule here

A fixed rule could prioritize pages using one or two thresholds, such as low recent traffic or old content. However, the dataset contains several interacting signals, including search demand, impressions, clicks, engagement, content age, freshness, and recent trends.

ML may help combine these signals and learn patterns that are difficult to capture with a single if-statement or manually chosen threshold. The goal is not to replace human judgment, but to produce a more useful ranking for decision-support.

The ML approach should be compared with a simple rule-based baseline to determine whether the added complexity is justified.

```python
print("Fixed rule: uses a small number of manually chosen thresholds.")
print("ML approach: combines multiple signals to produce a ranking.")
print("Goal: improve decision-support, not replace human review.")
```
